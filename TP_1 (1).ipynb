###Indice Masa Corporal

```python
#Funciones de agregación IMC
print("IMC")
print("="*60)

print(df["BMI"].describe())

# Histograma
print("\nIMC - Histograma")
print("="*60)
plt.figure(figsize=(5,5))

sns.histplot(
    df["BMI"],
    kde=True,
    )

plt.title("Distribución de IMC")
plt.show()
```

```python
#Segmentos por grupo de ICM y Riesgo de Diabetes
g0 = df[(df["BMI"] < 18) & (df["Diabetes_Risk"] == "High")] #IMC bajo y Riesgo Alto
g1 = df[(df["BMI"] > 18) & (df["BMI"] < 25) & (df["Diabetes_Risk"] == "High")] #IMC normal y Riesgo Alto
g2 = df[(df["BMI"] > 18) & (df["BMI"] < 25) & (df["Diabetes_Risk"] != "High")] #IMC normal y Riesgo Bajo o Moderado
g3 = df[(df["BMI"] >= 25) & (df["Diabetes_Risk"] == "High")] #IMC alto (sobrepeso) y Riesgo Alto

print("TAMAÑO DE LOS SEGMENTOS (IMC y Riesgo de Diabetes)")
print("=" * 60)
print(f"G0 Bajo Peso, Alto Riesgo: {len(g0):,}")
print(f"G1 Peso Normal, Alto Riesgo: {len(g1):,}")
print(f"G2 Peso Normal, Riesgo Normal: {len(g2):,}")
print(f"G3 Sobrepeso, Alto Riesgo: {len(g3):,}")
print()

#porcentaje perteneciente al cada Grupo
print("PORCENTAJE DE LOS SEGMENTOS (IMC y Riesgo de Diabetes)")
print("=" * 60)

porcentajeG0 = len(g0) / len(df) * 100
porcentajeG1 = len(g1) / len(df) * 100
porcentajeG2 = len(g2) / len(df) * 100
porcentajeG3 = len(g3) / len(df) * 100
print(
    f"G0 Bajo Peso, Alto Riesgo: {porcentajeG0:.2f}%"
    f"\nG1 Peso Normal, Alto Riesgo: {porcentajeG1:.2f}%"
    f"\nG2 Peso Normal, Riesgo Normal: {porcentajeG2:.2f}%"
    f"\nG3 Sobrepeso, Alto Riesgo: {porcentajeG3:.2f}%"
)


# Defino cuáles son los grupos de segmentos
grupos_glucosa = pd.concat([
    g0.assign(Grupo="G0"),
    g1.assign(Grupo="G1"),
    g2.assign(Grupo="G2"),
    g3.assign(Grupo="G3")
])


# Defino grupo con los porcentajes individuales
porcentajes_grupos = pd.Series({
    "G0": porcentajeG0,
    "G1": porcentajeG1,
    "G2": porcentajeG2,
    "G3": porcentajeG3
})


# GRAFICO DE BARRAS:
print("\nIMC - Grafico de Barras de Distribución Porcentual de Grupos")
print("="*60)

plt.figure(figsize=(5,5))
porcentajes_grupos.plot(kind="bar", color=["green", "orange", "yellow", "red"])

plt.title("Distribución Porcentual de los Grupos")
plt.ylabel("%")
plt.xlabel("Grupo")
plt.show()
```

```python
# BOXPLOT
print("\nIMC - Boxplot por Grupo")
print("="*60)
plt.figure(figsize=(5,5))


# Uso grupos_glucosa porque tiene definido los Grupo de cada fila
sns.boxplot(
    data=grupos_glucosa,
    x="Grupo",
    y="BMI",
      legend=False

)

plt.title("Distribución del IMC según Grupos")
plt.show()
```

```python
#Scatterplot IMC vs Hemoglobina Glicosilada (glucosa en sangre acumulada)

print("Scatterplot IMC vs Hemoglobina Glicosilada")
print("="*60)

sample = df.sample((5000), random_state=42)
#Sample permite elegir de manera aleatoria
#random_state semilla
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=sample,
    x="BMI",
    y="HbA1c",
    hue="Diabetes_Risk",
)
plt.title("IMC vs Glucosa en Sangre acumulada")
plt.tight_layout()
plt.show()
```

```python
#Histograma Glucosa en Sangre acumulada (HbA1c) por Grupo
print("Grupos de Glucosa")
print("=" * 60)
print(
    f"G0 Bajo Peso, Alto Riesgo"
    f"\nG1 Peso Normal, Alto Riesgo"
    f"\nG2 Peso Normal, Riesgo Normal"
    f"\nG3 Sobrepeso, Alto Riesgo"
)

print("\nGlucosa en Sangre acumulada por Grupos")
print("=" * 60)
plt.figure(figsize=(10,6))
sns.histplot(
    data=grupos_glucosa,
    x="HbA1c",
    hue="Grupo",
    kde=True,
    bins=30
)
plt.title("Distribución de HbA1c")
plt.tight_layout()
plt.show()

```

```python
# Defino grupos para las horas de sueño
df["Sleep_Group"] = pd.cut(
    #cut() sirve para dividir una variable numérica continua en intervalos
    df["Sleep_Hours"],
    bins=[0,6,8,24], #defino los límites de cada grupo
    labels=[
        "Poco Sueño", #0 a 6
        "Normal", #6 a 8
        "Mucho Sueño" #8 a 24
    ]
)

#Defino la tabla de datos a evaluar
print("\nHoras de Sueño por Riesgo de Diabetes")
print("="*60)
tabla_sleep = pd.crosstab(
    #crosstab() tabla de contingencia
    df["Sleep_Group"],
    df["Diabetes_Risk"],
    normalize="index"
)*100
print(tabla_sleep)

#Grafico de Barras horizontal
print("\nBarras - Horas de Sueño por Riesgo de Diabetes")
print("="*60)
tabla_sleep.plot(
    kind="barh",
    stacked=True,
    figsize=(5,5)
)
plt.title("Riesgo según Horas de Sueño")
plt.ylabel("%")
plt.show()
```

##Hipótesis y sus Conclusiones

###1- Obesidad

Las personas con mayor BMI (Indice Masa Corporal) presentan mayor riesgo de diabetes.

Variables:
*   BMI
*   Diabetes_Risk


CONCLUSION:
El boxplots principalmente muestran medianas claramente crecientes en el grupo con Mayor IMC.

Al mismo tiempo, en el Scatterplot se ve relación entre el IMC y la Hemoglobina Glicosilada, ya que los valores más altos, salvo casos aislados, de HbA1c se encuentran por encima de 6.


Esto muestra evidencias claras que el objetivo planteado es positivo al momento en que permiten asociar el Índice de Masa Corporal con el Riesgo de tener Diabetes

###2-Circunferencia de Cintura

Una mayor circunferencia de cintura aumenta el riesgo de diabetes.

Variables:
*   Waist_Circumference_cm
*   Diabetes_Risk


CONCLUSION:
La circunferencia de cintura presenta distribuciones muy similares entre los niveles de riesgo de diabetes. Las medianas, la dispersión y los rangos observados son prácticamente iguales, por lo que esta variable no es una hipótesis que permita poder evaluar el Riesgo de tener Diabetes.

###3-Antecedentes Familiares

Los pacientes con antecedentes familiares de diabetes presentan un riesgo de diabetes significativamente mayor que aquellos sin antecedentes familiares.

Variables:
*   Family_History_Diabetes
*   Diabetes_Risk


CONCLUSION:
Los antecedentes familiares de diabetes parecen estar asociados con una mayor probabilidad de pertenecer al grupo de riesgo alto de tener diabetes

El análisis de la distribución de antecedentes familiares por grupo, los grupos G0, G1 y G3, muestran que más del 55% de los individuos reportan antecedentes familiares de diabetes, mientras que en el grupo G2 el porcentaje disminuye hasta alrededor del 35%

Esta tendencia sugiere que el objetivo planteado es positivo respecto a que la presencia de antecedentes familiares esté asociados a un mayor riesgo de tener diabetes



###4-Sueño

Dormir menos horas se asocia con mayor riesgo de diabetes.

Variables:
*   Sleep_Hours
*   Diabetes_Risk


CONCLUSION:
Los 3 grupos de sueños definidos presentan riesgo alto en más de un 70% de la muestra
Por lo tanto, la cantidad de horas de sueño no es una variable que diferencie claramente los niveles de riesgo de diabetes.




###5-Edad

La probabilidad de riesgo alto aumenta con la edad.

Variables:
*   Age
*   Diabetes_Risk


CONCLUSION:
A medida que la edad aumenta, el riesgo de tener diabetes se incrementa, pasando de aproximadamente el 60% en menores de 30 años a más del 80% en mayores de 60 años.

Estos resultados sugieren que la edad constituye un factor positivo en la predisposición de tener alto Riesgo de diabetes.

#Casos de estudio (Insight):


Existen jóvenes (18-40 años) con Antecedentes Familiares y IMC normal o bajo (<25) que igualmente muestran un alto riesgo de tener diabetes, lo que indica que la predisposición genética podría compensar la ausencia de los factores tradicionales de riesgo asociados al sobrepeso.

Esto contradice la percepción habitual de que el alto riesgo de diabetes se concentra únicamente en personas con obesidad.

Este Insight apunta a un fenómeno de riesgo oculto.

Por lo tanto, los antecedentes familiares son un predictor clave para identificar población "aparentemente sana" que esta en situación de riesgo.




```python
#Filtrar la poblacion definida para el Insight
jovenes = df[
    (df["Age"] <= 30) &
    (df["BMI"] < 25)
]


# Genero una Tabla con los porcentajes
tabla = pd.crosstab(
    #crosstab() tabla de contingencia
    jovenes["Family_History_Diabetes"],
    jovenes["Diabetes_Risk"],
    normalize="index"
) * 100

# Ordenar las columnas
tabla = tabla[["Low", "Moderate", "High"]]

# Gráfico de Barra
print("\nInshigt")
print("="*60)
ax = tabla.plot(
    kind="bar",
    stacked=True,
    figsize=(5,5),
    color=["green", "yellow", "red"]
)

plt.title(
    "Riesgo Alto de Diabetes \n en Jóvenes (≤30 años)\n con IMC Normal o Bajo (<25)"
)
plt.xlabel("Antecedentes Familiares")
plt.ylabel("Porcentaje")
plt.legend(title="Diabetes Risk")

# Mostrar porcentajes
for c in ax.containers:
  #ax.containers almacena una lista con los objetos contenedores de las figuras geométricas
    ax.bar_label(c, fmt="%.1f%%", label_type="center", fontsize=8)
    #ax.bar_label para añadir etiquetas de texto con los valores numéricos
    #fmt="%.1f%%" lo deja en % con 1 solo decimal

plt.show()
```
